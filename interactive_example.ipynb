# An example of running the coE3T simulation interactively

This notebook loads the current client and launch utilities from the gitignored `METS-R_HPC/` asset. API responses use the native v2 schema: `messageType`, `status`, lowercase `data`, and camelCase fields.

# Section 1: Start, reset and terminate the simulation

```python
from pathlib import Path
import os
import shutil
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "METS-R_HPC":
    PROJECT_ROOT = PROJECT_ROOT.parent
METS_R_HPC_DIR = PROJECT_ROOT / "METS-R_HPC"
if not (METS_R_HPC_DIR / "clients" / "METSRClient.py").is_file():
    raise FileNotFoundError("Clone https://github.com/umnilab/METS-R_HPC.git into ./METS-R_HPC first")
shutil.copytree(PROJECT_ROOT / "data", METS_R_HPC_DIR / "data", dirs_exist_ok=True)

if str(METS_R_HPC_DIR) not in sys.path:
    sys.path.insert(0, str(METS_R_HPC_DIR))

from utils.util import *
from clients.METSRClient import METSRClient
clear_all()
```

```python
config = read_run_config(str(PROJECT_ROOT / "configs" / "run_interactive_nema.json"))
config.verbose = True # set to True to see the message
```

```python
# os.chdir("docker")
# os.system("docker-compose up -d")
# os.chdir("..")

# time.sleep(10) # wait 10s for the Kafka servers to be up

notebook_cwd = Path.cwd()
try:
    os.chdir(METS_R_HPC_DIR)
    sim_dirs = prepare_sim_dirs(config)
    run_simulation_in_docker(config)
    sim_dirs = [str(Path(sim_dir).resolve()) for sim_dir in sim_dirs]
    config.sim_dirs = sim_dirs
finally:
    os.chdir(notebook_cwd)

print(sim_dirs)
```

```python
client = METSRClient(
    host=config.metsr_host,
    port=config.ports[0],
    sim_folder=sim_dirs[0],
    config=config,
    verbose=False,
)
```

```python
# start the visualization server so the trajectory can be visualized at https://engineering.purdue.edu/HSEES/METSRVis/
client.start_viz()
```

```python
# Generating 500 trips in 5000 ticks.
for i in range(5000):
      if(i % 10 == 0):
            client.generate_trip(i, -1, -1)
      client.tick(1)
```

```python
client.reset()
```

```python
# This is doing the same thing as above, but in a more efficient way
for i in range(500):
      client.generate_trip(i, -1, -1)
      client.tick(10)
```

```python
# one can close the connection without terminating the simulation
client.close()

# and reestablish the connection ideally from the other machine
client = METSRClient(
    host=config.metsr_host,
    port=config.ports[0],
    sim_folder=sim_dirs[0],
    config=config,
    verbose=True,
)
```

```python
client.reset()
```

```python
# generate trips between roads
import random
road_ids = client.query_road()["roadIds"]
road_ids[:5]
```

```python
client.start_viz()
```

```python
for i in range(500):
      client.tick(10)
      origin_road = random.choice(road_ids)
      destination_road = random.choice(road_ids)
      client.generate_trip_between_roads(i, origin_road, destination_road)
```

```python
# terminate the simulation
client.terminate()
```

# Section 2: Query APIs

In this example, we will see how to use query to extract data from the simulation.

```python
from pathlib import Path
import os
import shutil
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "METS-R_HPC":
    PROJECT_ROOT = PROJECT_ROOT.parent
METS_R_HPC_DIR = PROJECT_ROOT / "METS-R_HPC"
if not (METS_R_HPC_DIR / "clients" / "METSRClient.py").is_file():
    raise FileNotFoundError("Clone https://github.com/umnilab/METS-R_HPC.git into ./METS-R_HPC first")
shutil.copytree(PROJECT_ROOT / "data", METS_R_HPC_DIR / "data", dirs_exist_ok=True)

if str(METS_R_HPC_DIR) not in sys.path:
    sys.path.insert(0, str(METS_R_HPC_DIR))

from utils.util import *
from clients.METSRClient import METSRClient
```

```python
config = read_run_config(str(PROJECT_ROOT / "configs" / "run_interactive_Birmingham.json"))
config.verbose = True # set to True to see the message
```

```python
notebook_cwd = Path.cwd()
try:
    os.chdir(METS_R_HPC_DIR)
    sim_dirs = prepare_sim_dirs(config)
    run_simulation_in_docker(config)
    sim_dirs = [str(Path(sim_dir).resolve()) for sim_dir in sim_dirs]
    config.sim_dirs = sim_dirs
finally:
    os.chdir(notebook_cwd)

print(sim_dirs)
```

```python
# If no metsr_port is specified, the runner will find the port automatically 
config.ports
```

```python
# Better to  set up a larger timeout (default is 30s) if your map/vehicle number is huge
sim_client = METSRClient(
    host=config.metsr_host,
    port=config.ports[0],
    sim_folder=sim_dirs[0],
    config=config,
    verbose=False,
    timeout=300,
)
```

```python
sim_client.start_viz()
```

```python
for i in range(500):
      sim_client.generate_trip(i, -1, -1)
      sim_client.tick(10)
```

```python
zones = sim_client.query_zone()["zoneIds"]
```

```python
import numpy as np

for i in range(60):
      for j in range(50):
            sim_client.add_taxi_requests(int(np.random.choice(zones)), int(np.random.choice(zones)), 1)
      sim_client.tick(300, wait_forever = True)
      sim_client.render()

for i in range(60):
      sim_client.tick(300, wait_forever = True)
      sim_client.render()
```

Now let's do some query.

```python
# With no ID, v2 returns the fleet index in the top-level busIds field.
sim_client.query_bus()
```

```python
# Taxi fleet queries use the analogous top-level taxiIds field.
sim_client.query_taxi()
```

```python
# This vehicle is in state 0, meaning it is parking
sim_client.query_taxi(0)
```

```python
# Facility index queries expose canonical camelCase ID fields.
sim_client.query_road()["roadIds"]
```

```python
# This id can be found in the road csv (or SUMO XML) in the data folder
sim_client.query_road('-1004394328#10')
```

```python
# Also, a list of ID can be used to query multiple roads
sim_client.query_road(sim_client.query_road()["roadIds"][:5])
```

```python
sim_client.query_signal()
```

```python
sim_client.query_signal(2)
```

```python
sim_client.query_chargingStation()
```

```python
sim_client.query_chargingStation(-10)
```

```python
sim_client.terminate()
```

# Section 3: Control APIs

In this section, we will see how to interact with the simulation.

```python
from pathlib import Path
import os
import shutil
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == "METS-R_HPC":
    PROJECT_ROOT = PROJECT_ROOT.parent
METS_R_HPC_DIR = PROJECT_ROOT / "METS-R_HPC"
if not (METS_R_HPC_DIR / "clients" / "METSRClient.py").is_file():
    raise FileNotFoundError("Clone https://github.com/umnilab/METS-R_HPC.git into ./METS-R_HPC first")
shutil.copytree(PROJECT_ROOT / "data", METS_R_HPC_DIR / "data", dirs_exist_ok=True)

if str(METS_R_HPC_DIR) not in sys.path:
    sys.path.insert(0, str(METS_R_HPC_DIR))

from utils.util import *
from clients.METSRClient import METSRClient
```

```python
config = read_run_config(str(PROJECT_ROOT / "configs" / "run_interactive_Birmingham.json"))
config.verbose = True # set to True to see the message
```

```python
notebook_cwd = Path.cwd()
try:
    os.chdir(METS_R_HPC_DIR)
    sim_dirs = prepare_sim_dirs(config)
    run_simulation_in_docker(config)
    sim_dirs = [str(Path(sim_dir).resolve()) for sim_dir in sim_dirs]
    config.sim_dirs = sim_dirs
finally:
    os.chdir(notebook_cwd)

print(sim_dirs)
```

```python
client = METSRClient(
    host=config.metsr_host,
    port=config.ports[0],
    sim_folder=sim_dirs[0],
    config=config,
    verbose=False,
    timeout=300,
)
```

```python
# no private vehicle is in the simulation yet
client.query_vehicle()["privateVehicleIds"]
```

```python
# generate a private vehicle trip
client.generate_trip(1, -1, -1)
```

```python
# state 8 means the vehicle is performing a private trip
client.query_vehicle(1, private_veh = True)
```

```python
# can also generate a private vehicle trip between two roads
import random
road_ids = client.query_road()["roadIds"]

for i in range(1, 101):
      origin_road = random.choice(road_ids)
      destination_road = random.choice(road_ids)
      client.generate_trip_between_roads(i, origin_road, destination_road)
```

```python
client.query_vehicle(100, private_veh = True)
```

```python
client.query_zone()
```

```python
# start the visualization server so the trajectory can be visualized at https://engineering.purdue.edu/HSEES/METSRVis/
client.start_viz()
```

```python
client.terminate()
```
